```python
from langchain_community.document_loaders import TextLoader
```

```python
loader = TextLoader("./README.md")
doc = loader.load()
```

```python
# Entire text file is loaded as one document
len(doc)
```

```python
type(doc[0])
```

```python
# See first 100 characters fom document
print(doc[0].page_content[:100])
```

```python
# It is a text loader but works better with MD files. mode='single' mean read entire file as one document.
from langchain_community.document_loaders import UnstructuredMarkdownLoader
```

```python
import nltk
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
```

```python
loader = UnstructuredMarkdownLoader("./README.md", mode='single')
docs = loader.load()
```

```python
len(docs)
```

```python
type(docs[0])
```

```python
print(docs[0].page_content[:100])
```

```python
# mode="elements" mean create a seperate node for each document.
from langchain_community.document_loaders import UnstructuredMarkdownLoader

loader = UnstructuredMarkdownLoader("./README.md", mode="elements")
docs = loader.load()
```
