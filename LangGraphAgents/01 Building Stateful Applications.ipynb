Create Graph in LangGraph

1- Define Schema/ graph state

```python
# State can be defined using TypedDict
from typing import TypedDict
```

```python
# Lets create our state which has 2 attributes
class State_with_attribute(TypedDict):
    attribute1: str
    attribute2: str
```

2-Define node functionality

```python
# Lets create a function (we will add this to a node) which take out initial state and update "attribute1" and it return the updated state.
def Graph_function(state: State_with_attribute) -> State_with_attribute:
    state['attribute1'] = 'Value updated by node'
    return state
```

3- Build Graph with state

```python
# Now lets initialize the graph
from langgraph.graph import StateGraph
# Add our state to graph
graph = StateGraph(State_with_attribute)
# Lets add first node to graph, and this node will execute "Graph_function" function
graph.add_node('node1', Graph_function)
```

4- Connect Nodes

```python
from langgraph.graph.state import END, START
# We can do this in two ways:
graph.add_edge(START, 'node1')

graph.add_edge('node1', END)
```

5- Complile Grpah

```python
compiled_graph = graph.compile()
```

```python
from IPython.display import Image
Image(compiled_graph.get_graph().draw_mermaid_png())
```

6- Invoke grpah

```python
# Lets invoke the graph with our own defined state and graph will update the state,
compiled_graph.invoke({"attribute1": "Initial value",
                       "attribute2": "Initial value"})
```
