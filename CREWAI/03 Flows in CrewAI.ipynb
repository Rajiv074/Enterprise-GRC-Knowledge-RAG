# Methods- @Start and @Listen

```python
import nest_asyncio
nest_asyncio.apply()
```

```python
import random
from crewai.flow.flow import Flow, listen, router, start, and_
from pydantic import BaseModel
```

```python
class flow_state(BaseModel):
    a: int = 2
    b: int = 3
    sum_of_numbers: int =0
```

```python
class crew_flow(Flow[flow_state]):
    @start()
    def get_number(self):
        print("Getting number for addition")
        self.state.a = 2
        self.state.b = 3
        return flow_state
    
    @listen(get_number)
    def sum_numbers(self):
        print("Sum of numbers is")
        self.state.sum_of_numbers = self.state.a + self.state.b
        print(self.state.sum_of_numbers)
        return flow_state
        
```

```python
flow = crew_flow()
flow.kickoff()
```

 # Method-  And_

```python
class crew_flow(Flow[flow_state]):
    @start()
    def get_first_numbers(self):
        print("Getting first number for addition")
        self.state.a=2
        return flow_state

    @listen(get_first_numbers)
    def get_second_numbers(self):
        print("Getting second number for addition")
        self.state.b=3
        return flow_state
        
    @listen(and_(get_first_numbers, get_second_numbers))
    def sum_numbers(self):
        print("Sum of numbers")
        self.state.sum_of_numbers=self.state.a+self.state.b
        print(self.state.sum_of_numbers)
        return flow_state
```

```python
flow = crew_flow()
flow.kickoff()
```

# @Router

```python
class CrewFlowState(BaseModel):
    Query_Type: str = "NA"

class Crew_Flow(Flow[CrewFlowState]):

    @start()
    def Query_Classification(self):
        print("Question is related to ->")
        self.state.Query_Type = "HR"
        print(self.state.Query_Type)

    @router(Query_Classification)
    def check_outcome(self):
        if self.state.Query_Type == "Billing":
            return "Billing_Processes"
        else:
            return "HR_Processes"

    @listen("Billing_Processes")
    def Billing_flow(self):
        print("Billing Processes Initiated!")

    @listen("HR_Processes")
    def HR_Flow(self):
        print("HR Processes Initiated!")
```

```python
flow = Crew_Flow()
flow.kickoff()
```
