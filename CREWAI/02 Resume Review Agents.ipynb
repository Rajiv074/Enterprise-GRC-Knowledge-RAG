```python
Technical_reviewer = Agent(
    llm=llm,
    role="ATS Resume Reviewer",
    goal="Evaluate resume against job description",
    backstory="""Expert in ATS systems and resume keyword optimization.""",
    verbose=True
)
```

```python
formating_reviewer = Agent(
    llm=llm,
    role="HR Resume Reviewer",
    goal="Review resume professionally",
    backstory="""HR expert with experience in hiring software engineers.""",
    verbose=True
)
```

```python
technical_task = Task(
    description=f"""
    Compare the resume with the job description.

    Resume:
    {resume_text}

    Job Description:
    {job_description}

    Provide:
    1. ATS score
    2. Missing keywords
    3. Improvement suggestions
    """,

    expected_output="""
    A detailed ATS evaluation report including:
    - ATS score
    - Missing keywords
    - Resume improvement suggestions
    """,

    agent=Technical_reviewer
)
```

```python
# Task 2
formating_task = Task(
    description=f"""
    Review the resume professionally.

    Resume:
    {resume_text}

    Provide:
    1. Grammar improvements
    2. Formatting suggestions
    3. Professional summary feedback
    """,

    expected_output="""
    A professional resume review report including:
    - Grammar corrections
    - Formatting improvements
    - Professional summary feedback
    - Overall resume enhancement suggestions
    """,

    agent=formating_reviewer
)
```

```python
crew = Crew(
    agents=[Technical_reviewer, formating_reviewer],
    tasks=[technical_task, formating_task],
    planning=True,
    verbose=True
)
```

```python
output=crew.kickoff()
```
