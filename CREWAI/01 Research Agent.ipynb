We wnat an agentic system which can generate 5 positioning ideas for a given product. We will create 2 agents:

Market Research Agent: A market research agent which is going to get the relevant information from internet.

Product Strategy: It will analyze the data from Market Research agent and generate positioning Strategies (Targeted audience, Market Analysis)

```python
# Setup your LLM keys
import os
from dotenv import load_dotenv, find_dotenv
load_dotenv()
```

```python
OPENAI_API_KEY = os.environ["OPENAI_API_KEY"]
SERPER_API_KEY = os.environ["SERPER_API_KEY"]
```

```python
from crewai import Task, Agent, LLM, Crew
from crewai_tools import SerperDevTool
```

```python
llm = LLM(model="gpt-4o-mini")
```

```python
product_name = "dishwasher"
```

```python
market_researcher = Agent(
    role="Market Researcher",
    goal="Analyze market trends for the product launch",
    backstory="Experinced in market trends and consumer behaviour analysis",
    tools=[SerperDevTool()],
    verbose=True
)
```

```python
strategist_backstory = "marketing strategy"
```

```python
product_strategist = Agent(
    role="Product Strategist",
    goal="Create effective positioning product Strategies for the product",
    backstory=f"Skilled in competitive positioning and {strategist_backstory}",
    verbose=True
)
```

```python
gather_market_insights_task = Task(
    description=f"Browse the internet to gather insights on current market trends for the launch of the {product_name} product.",
    expected_output=f"List of relevant market trends and consumer preferences, relevant to {product_name}",
    agent=market_researcher,
)
```

```python
develop_positioning_strategy_task = Task(
    description="Based on the market insights,"
                f"create a positioning strategy for the {product_name} product, including analysis for impact and target audience.",
    expected_output="A positioning strategy with target audience and impact notes",
    agent=product_strategist,
    verbose=True,
)
```

```python
crew = Crew(
    agents=[market_researcher,product_strategist],
    tasks=[gather_market_insights_task,develop_positioning_strategy_task],
    planning=True
    
)
```

```python
# gather_market_insights_task task will be executed by market_researcher and output of market_researcher will got to Product_strategist which will
# execute develop_positioning_strategy_task task.

# First it will pick the ficrt taks to execute, once task1 is executed crew will start executing the next tasks in the crew.

output=crew.kickoff()
```

```python
print(output.raw)
```

# Stractured Tasks output

The agent should draft a Leave Policy document and return a structured object containing sections such as:
- Policy Title
- Introduction
- Purpose
- Scope
- Leave Entitlements
- Approval Process
- Responsibilities
- Conclusion

```python
from pydantic import BaseModel

class LeavePolicy(BaseModel):
    policy_title: str
    introduction: str
    purpose: str
    scope: str
    leave_entitlements: str
    approval_process: str
    responsibilities: str
    conclusion: str
```

```python
policy_writer = Agent(
    role="HR Policy Writer",
    goal="Create professional HR policies",
    backstory="""
    You are an experienced HR policy consultant.
    You create well-structured and professional policy documents.
    """,
    verbose=True,
    llm=llm
)
```

```python
leave_policy_task = Task(
    description="""
    Create a comprehensive Employee Leave Policy.

    The policy should include:

    - Policy Title
    - Introduction
    - Purpose
    - Scope
    - Leave Entitlements
    - Approval Process
    - Responsibilities
    - Conclusion

    The policy should be suitable for a medium-sized organization.
    """,
    expected_output="A structured leave policy document",
    output_pydantic=LeavePolicy,
    agent=policy_writer
)
```

```python
crew = Crew(
    agents=[policy_writer],
    tasks=[leave_policy_task],
    verbose=True
)
```

```python
result = crew.kickoff()
```

```python
policy = leave_policy_task.output.pydantic

print(policy)
```

```python
print(policy.policy_title)
```
