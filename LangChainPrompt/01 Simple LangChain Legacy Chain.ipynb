```python
api_key = os.getenv('OPENAI_API_KEY')
TAVILY_API_KEY = os.getenv('TAVILY_API_KEY')
print(f"TAVILY_API_KEY: {TAVILY_API_KEY}")
```

```python
if not api_key:
    print("No API key was found!")
elif not api_key.startswith("sk-proj"):
    print("An API key was found, but it doesn't start sk-proj-;")
elif api_key.strip() != api_key:
    print("An API key was found, but it looks like it might have space or tab characters at the start or end")
else:
    print("API key found and looks good so far!")
```

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model_name="gpt-4o-mini", temperature=0)
```

```python
from langchain_core.prompts import ChatPromptTemplate

# create a prompt template to accept user queries
prompt_txt = "Explain {topic} in 1 line"
prompt_template = ChatPromptTemplate.from_template(prompt_txt)
```

```python
llmchain = (prompt_template
              |
           llm)
```

```python
response = llmchain.invoke({'topic' : 'Cricket'})
print(response.content)
```

```python
response = llmchain.invoke({'topic' : 'machine learning'})
print(response.content)
```
