Exploring Built-in Tools

```python
from langchain_community.tools import WikipediaQueryRun

from langchain_community.utilities import WikipediaAPIWrapper

# WikipediaAPIWrapper is API to search on wikipedia. top_k_results=3 mean it will return top 3 results
wiki_api_wrapper = WikipediaAPIWrapper(top_k_results=3, doc_content_chars_max=8000)

#WikipediaQueryRun is wrapper to run wikipedia API
wiki_tool = WikipediaQueryRun(api_wrapper=wiki_api_wrapper, features="lxml")
```

```python
wiki_tool.description
```

```python
# This tool has only one parameter "query"
wiki_tool.args
```

```python
print(wiki_tool.invoke({"query": "AI Agents"}))
```

You can customize the default tool with its own name, description and so on as follows

```python

from langchain_core.tools import Tool

wiki_tool_init = Tool(name="Wikipedia",
                      func=wiki_api_wrapper.run,  # wiki_api_wrapper.run is wikipedia api build in LangChain
                      description="useful when you need a detailed answer about any general topic")
```

```python
wiki_tool_init.description
```

```python
wiki_tool_init.args
```

```python
print(wiki_tool_init.invoke({"tool_input": "Machine Learning"}))
```

Implement custom tools in LangChain

A tool is consist of:

Tool Name
Tool description- description of what the tool is
Input parameter details- JSON schema of what the inputs to the tool are
Actual functionality of tool- The function to call
Whether the result of a tool should be returned directly to the user
The name, description, and JSON schema helpss the LLM in determining what action to take, while the corresponding function represents the actual execution of that action

Building multiplication tool

```python
from langchain_core.tools import tool

# tool is a decorator, which will automatically generate the tool name, tool descreption, input schema etc.

@tool
def multiply(a, b):
    """Multiply two numbers."""
    return a * b
```

```python
print(multiply.name)
```

```python
print(multiply.description)
```

```python
print(multiply.args)
```

```python
multiply.invoke({"a": 2.1, "b": 3.2})
```
